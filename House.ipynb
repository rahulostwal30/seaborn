feature scaling

from sklearn.preprocessing import StandardScaler
standard = StandardScaler()
df[['Area','Year']] = standard.fit_transform(df[['Area','Year']])

encoding

```python
from sklearn.preprocessing import OneHotEncoder
oneencode = OneHotEncoder(sparse_output=False)
one_encoding= oneencode.fit_transform(df[['Location']])
encode_df = pd.DataFrame(one_encoding,
                         columns=oneencode.get_feature_names_out(),
                         index=df.index)
df.drop('Location', axis=1, inplace=True)
df[encode_df.columns] = encode_df
```

```python
encode_df
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordencode = OrdinalEncoder(categories=[['Poor', 'Fair', 'Good', 'Excellent']])
ord_encoding = ordencode.fit_transform(df[['Condition']])

ord_encode_df = pd.DataFrame(
    ord_encoding,
    columns=['Condition'],
    index=df.index
)

df.drop('Condition', axis=1, inplace=True)
df[ord_encode_df.columns] = ord_encode_df.astype(int)
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['No', 'Yes']])

df['Garage'] = encoder.fit_transform(df[['Garage']]).astype(int)
```

```python
df
```

```python
x = df.drop('Price',axis=1)
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

Model train

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)

r2 = r2_score(y_test, y_pred)
print("R2 Score:", r2)
```
